```python
lst1 = [6,2,3,5,2,7,1]
# Expected output: 1

min = lst1[0]

for i in lst1:
  if i < min:
    min = i
print("Minimum value is {0}".format(min))

# min = lst1[0]
# for i in range(1,len(lst1)):
#   if lst1[i] < min:
#     min = lst1[i]
# print("Minimum value is {0}".format(min))
```

2. Sorting a list

```python
# Selection sort

lst1 = [6,2,3,5,1,2,7]
n = len(lst1)

for i in range(n-1):
  min_idx = i
  for j in range(i+1, n):
    if lst1[j] < lst1[min_idx]:
      min_idx = j
  temp = lst1[min_idx]
  lst1[min_idx] = lst1[i]
  lst1[i] = temp

print("Selection sort output:")
print(lst1)

# Bubble sort

lst2 = [5,9,2,1,7,6,4]
n = len(lst2)

for i in range(n-1):
  for j in range(n-i -1):
    if lst2[j] > lst2[j + 1]:
      lst2[j], lst2[j + 1] = lst2[j + 1], lst2[j]

print("\nBubble sort output:")
print(lst2)
```

3. Find second largest element in a list.

```python
arr = [5,4,6,2,-1,8]
# Sorted list: [-1,2,4,5,6,8]
# Expected output: 6

l1 = arr[0]
l2 = arr[0]

for i in arr:
  if i > l1:
    l2 = l1
    l1 = i
  elif i > l2:
    l2 = i

print("Second largest element: {0}".format(l2))
```

5. Move zeros to the end

```python
l1 = [0,2,0,5,4,0]

# Assume 0 is the position where first non-zero number can come
idx = 0

# Move all non-zero elements to the beginning
for i in range(len(l1)):
  if l1[i] != 0:
    l1[idx] = l1[i]
    idx += 1

# Assign 0 to remaining positions
for j in range(idx,len(l1)):
  l1[j] = 0

print(l1)
```

7. Reverse a list

```python
lst = [1,2,3,4,6,7]

l = 0
r = len(lst) - 1

while l < r:
  lst[l], lst[r] = lst[r], lst[l]
  l += 1
  r -= 1

print(lst)

# Using list slicing
print(lst[::-1])
```

```python
# Using in-built function

lst.reverse()
print(lst)

#using slicing method
print(lst[::-1])
```

**2 Sum Problem**

```python
arr = [2, 7, 11, 15]
target = 9

def two_sum_brute(arr, target):
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] + arr[j] == target:
                return [i, j]
    return []
print("Brute:", two_sum_brute(arr, target))
```
