```python
from pyspark.sql import SparkSession
import random

# Initialize Spark session
spark = SparkSession.builder.appName("JoinPractice").getOrCreate()

# -----------------------
# Employees DataFrame
# -----------------------
departments = ["HR", "IT", "Finance", "Marketing", "Sales", "Support"]
employee_data = []
for i in range(1, 101):
    employee_data.append((
        i,
        f"Employee_{i}",
        random.choice(departments),
        random.randint(1, 10) if i > 10 else None  # Random manager_id or None
    ))

employees_df = spark.createDataFrame(employee_data, ["emp_id", "name", "dept", "manager_id"])

# -----------------------
# Departments DataFrame
# -----------------------
locations = ["New York", "San Francisco", "Chicago", "Houston", "Austin"]
dept_data = []

# Matching departments
for dept in departments:
    dept_data.append((dept, random.choice(locations)))

# Additional unmatched departments
for i in range(6, 11):
    dept_data.append((f"Dept_{i}", random.choice(locations)))

departments_df = spark.createDataFrame(dept_data, ["dept", "location"])

# -----------------------
# Salaries DataFrame
# -----------------------
years = [2022, 2023, 2024]
salary_data = []
for i in range(1, 101):
    salary_data.append((
        i,
        random.choice(years),
        round(random.uniform(50000, 150000), 2)
    ))

salaries_df = spark.createDataFrame(salary_data, ["emp_id", "year", "salary"])

# -----------------------
# Preview the Data
# -----------------------
print("Employees:")
employees_df.show(5)

print("Departments:")
departments_df.show(5)

print("Salaries:")
salaries_df.show(5)
```
